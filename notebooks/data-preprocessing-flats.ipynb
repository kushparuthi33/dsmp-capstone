## CLAUDE CLEANING

```python
# continue from my cleaned df (on a copy) and bring back the raw file for the columns that need re-parsing
import ast
raw = pd.read_excel('/Users/mishakush/Desktop/Capstone_Project_DSMP1/data/raw/flats.xlsx').loc[df.index]
cdf = df.copy()
cdf.shape
```

```python
# 1. duplicate listings: same property_id scraped twice (df.duplicated() gives 0 only because the rating list order differs) -> keep the latest copy
cdf['property_id'] = raw['property_id']
print(cdf['property_id'].duplicated().sum())
cdf = cdf[~cdf['property_id'].duplicated(keep='last')]
# 2. one listing has no price at all -> useless as a target
cdf = cdf[cdf['price'].notnull()]
cdf.shape
```

```python
# 3. society: redo from raw so only the rating (d.d ★) is removed, not digits of the name (dlf new town heights 1 / 2, godrej 101 ...)
def clean_society(name):
    if pd.isnull(name):
        return 'independent'
    name = re.sub(r'\d\.\d\s?★', '', name)
    return re.sub(r'\s+', ' ', name).strip().lower().replace('appartment', 'apartment')

cdf['society'] = raw['society'].apply(clean_society)
# same society written with different spacing / hyphens -> most common spelling
key = cdf['society'].str.replace(r'[^a-z0-9]', '', regex=True)
cdf['society'] = cdf.groupby(key)['society'].transform(lambda s: s.mode()[0])
cdf['society'] = cdf['society'].replace({
    'ansals api esencia': 'ansal api esencia', 'ansal height 86': 'ansal heights 86',
    'breeze global heights': 'breez global heights', 'essel tower': 'essel towers',
    'sidhivinayak apartments': 'siddhi vinayak apartments',
    'ss': 'ss the leaf',   # sector 85 listings with the same flat sizes as ss the leaf
    '.': 'independent', 'nul': 'independent', 'not applicable': 'independent',
    'apartment': 'independent', 'builder floor': 'independent', 'rwa': 'independent'})
cdf['society'].nunique()
```

```python
# 4. sector / locality from property_name
def get_sector(name):
    loc = name.split(' in ', 1)[1].lower()
    m = re.search(r'sector[\s-]*(\d+)\s*([a-z])?\b', loc)
    if m is None:
        return re.sub(r'\s+', ' ', re.sub(r'(?<!new) gurgaon$', '', loc)).strip()
    sector = 'sector ' + m.group(1) + (m.group(2) or '')
    if 'sohna' in loc:
        sector += ' sohna'
    elif 'manesar' in loc:
        sector += ' imt manesar'
    return sector

cdf.insert(loc=3, column='sector', value=cdf['property_name'].apply(get_sector))
cdf['sector'].nunique()
```

```python
# 5. price: treat_price rounded Lac prices to 2 decimals (17.5 Lac -> 0.18, 24.5 Lac -> 0.24) -> keep the exact value in crore
def price_in_crore(x):
    value, unit = x.split(' ')
    return round(float(value) / 100, 4) if unit == 'Lac' else float(value)

cdf['price'] = raw['price'].loc[cdf.index].apply(price_in_crore)
```

```python
# 6. areaWithType -> 3 numeric columns in sq.ft., fixing the unit / typing mistakes on the way
def parse_areas(row):
    out = {'area_fixed': False}
    labels = {'super_built_up_area': 'Super Built up area', 'built_up_area': 'Built Up area', 'carpet_area': 'Carpet area'}
    for col, label in labels.items():
        m = re.search(label + r':? ?([\d.]+)(?: sq\.ft\.)? ?(?:\(([\d.]+) sq\.m\.\))?', row['areaWithType'])
        if m is None:
            out[col] = np.nan
            continue
        value = float(m.group(1))
        if m.group(2) is None:
            # no (sq.m.) conversion shown -> the number itself was entered in sq.m.
            area = value * 10.7639
            if row['price'] * 10000000 / area < 2000:      # ...unless that gives an impossible rate: it was sq.ft. all along
                area = value
                out['area_fixed'] = True
        else:
            sqm = float(m.group(2))
            # number not in sq.ft. (sq.yd.) -> convert from the sq.m. figure
            area = value if abs(value / sqm - 10.7639) < 0.1 else sqm * 10.7639
            while row['price'] * 10000000 / area < 2000:    # decimal point lost: 569243 is really 569.243
                area = area / 10
                out['area_fixed'] = True
        if area < 100 or row['price'] * 10000000 / area > 100000:   # a zero lost: 72 sq.ft. flat, 4 BHK of 700 sq.ft. for 14 crore
            area = area * 10
            out['area_fixed'] = True
        out[col] = round(area, 2)
    return pd.Series(out)

areas = cdf.apply(parse_areas, axis=1)
for i, col in enumerate(['super_built_up_area', 'built_up_area', 'carpet_area']):
    cdf.insert(loc=8 + i, column=col, value=areas[col].astype(float))
areas['area_fixed'].sum()
```

```python
# 7. area (my column = price / price_per_sqft) is only right when the site's rate is: the rate is stale where the price was changed
#    after posting, and wrong where the area was wrong -> if it matches none of the listed areas, take the listed area and redo the rate
implied = cdf['price'] * 10000000 / cdf['price_per_sqft']
listed = cdf[['super_built_up_area', 'built_up_area', 'carpet_area']]
matches = listed.sub(implied, axis=0).abs().div(implied, axis=0).lt(0.05).any(axis=1)
listed_area = cdf['super_built_up_area'].fillna(cdf['built_up_area']).fillna(cdf['carpet_area'])
cdf['area'] = round(implied.where(matches, listed_area))
cdf.loc[~matches, 'price_per_sqft'] = round(cdf['price'] * 10000000 / cdf['area'])
(~matches).sum()
```

```python
# 8. floors: floorNum as a number + total floors of the building (was thrown away)
cdf['floorNum'] = cdf['floorNum'].astype(float).astype('Int64')
total_floors = raw['floorNum'].str.extract(r'of (\d+) Floors')[0].astype(float).astype('Int64')
cdf.insert(loc=cdf.columns.get_loc('floorNum') + 1, column='total_floors', value=total_floors)
```

```python
# 9. facing: 'NA' is read back as NaN by pd.read_csv -> use 'unknown'
cdf['facing'] = cdf['facing'].replace('NA', 'unknown')
# 10. agePossession: dates (Excel turned 'Dec 2023' into datetimes), 'By 2024', 'Within 6 months' all mean not ready yet
def clean_age(x):
    if pd.isnull(x) or x == 'undefined':
        return 'unknown'
    if 'Year Old' in str(x):
        return x
    return 'Under Construction'

cdf['agePossession'] = cdf['agePossession'].apply(clean_age)
cdf['agePossession'].value_counts()
```

```python
# 11. additionalRoom: same rooms in a different order were different categories
cdf['additionalRoom'] = cdf['additionalRoom'].apply(lambda x: ', '.join(sorted(x.lower().split(','))))
cdf['additionalRoom'].nunique()
```

```python
# 12. address: remove flat / unit number at the start, extra spaces, and fill the missing ones from property_name
def clean_address(row):
    address = row['address']
    if pd.isnull(address):
        return row['property_name'].split(' in ', 1)[1] + ', Gurgaon, Haryana'
    parts = [re.sub(r'\s+', ' ', p).strip() for p in address.split(',')]
    while len(parts) > 3 and re.fullmatch(r'[\w/-]*\d[\w/-]*', parts[0]) and not parts[0].lower().startswith('sector'):
        parts = parts[1:]
    return ', '.join(p for p in parts if p)

cdf['address'] = cdf.apply(clean_address, axis=1)
```

```python
# 13. list columns: missing -> [], extra spaces / repeated items removed, rating sorted so the same rating always looks the same
def clean_list(x, sort=False):
    if pd.isnull(x):
        return '[]'
    items = [re.sub(r'\s+', ' ', i).strip() for i in ast.literal_eval(x)]
    items = list(dict.fromkeys(i for i in items if i))
    return str(sorted(items) if sort else items)

for col in ['nearbyLocations', 'furnishDetails', 'features']:
    cdf[col] = cdf[col].apply(clean_list)
cdf['rating'] = cdf['rating'].apply(clean_list, sort=True)
```

```python
# 14. nearbyLocations: same place written in different ways -> one name
places = cdf['nearbyLocations'].apply(ast.literal_eval)
alias = {'indira gandhi intl airport': 'Indira Gandhi International Airport', 'igi airport': 'Indira Gandhi International Airport',
         'dwarka expy': 'Dwarka Expressway'}
all_places = places.explode().dropna()
common = all_places.groupby(all_places.str.lower()).agg(lambda v: v.mode()[0]).to_dict()
cdf['nearbyLocations'] = places.apply(lambda l: str(list(dict.fromkeys(alias.get(p.lower(), common[p.lower()]) for p in l))))
```

```python
# 15. description: line breaks / repeated spaces -> single space (keeps one listing on one csv line)
cdf['description'] = cdf['description'].str.replace(r'\s+', ' ', regex=True).str.strip()
```

```python
cdf = cdf.drop(columns=['property_id']).reset_index(drop=True)
cdf.info()
```
